## 3.1 CLASSES, `__init__`, AND METHODS

A class is a blueprint for creating objects.
- `__init__` is the constructor — called automatically when creating an object.
- `self` refers to the current instance of the class.
- Class variables are shared; instance variables belong to each object.

```python
class Employee:
    # Class variable — shared by ALL instances
    company = "TechCorp"
    employee_count = 0

    # Constructor — runs when you create an object
    def __init__(self, name: str, salary: float):
        self.name = name              # Instance variable — unique to each object
        self.salary = salary
        Employee.employee_count += 1  # Modify class variable

    # Instance method — operates on the specific object
    def display(self):
        print(f"  {self.name} | ${self.salary:,.0f} | {self.company}")

    # Instance method — business logic
    def annual_salary(self) -> float:
        return self.salary * 12

    # __str__ — human-readable string representation
    def __str__(self):
        return f"Employee({self.name}, ${self.salary:,.0f})"

    # __repr__ — developer-friendly representation
    def __repr__(self):
        return f"Employee(name={self.name!r}, salary={self.salary})"

print("--- Creating Objects ---")
emp1 = Employee("Alice", 8000)
emp2 = Employee("Bob", 9500)

emp1.display()
emp2.display()

print(f"\n  str(emp1):  {str(emp1)}")
print(f"  repr(emp1): {repr(emp1)}")
print(f"  Annual salary: ${emp1.annual_salary():,.0f}")
print(f"  Total employees: {Employee.employee_count}")

# Class variable vs Instance variable
print("\n--- Class vs Instance Variable ---")
print(f"  emp1.company = {emp1.company}")
print(f"  emp2.company = {emp2.company}")
Employee.company = "NewCorp"          # Change for ALL instances
print(f"  After changing class var:")
print(f"  emp1.company = {emp1.company}")
print(f"  emp2.company = {emp2.company}")
```

## 3.2 INHERITANCE

Inheritance lets a child class reuse code from a parent class.
- The child can override methods to customize behavior.
- `super()` calls the parent's method.
- Polymorphism: same method name, different behavior per class.

```python
class Animal:
    def __init__(self, name: str, sound: str = "..."):
        self.name = name
        self.sound = sound

    def speak(self) -> str:
        return f"{self.name} says {self.sound}"

    def info(self) -> str:
        return f"{self.__class__.__name__}(name={self.name!r})"

# Dog inherits from Animal
class Dog(Animal):
    def __init__(self, name: str, breed: str):
        super().__init__(name, sound="Woof!")  # Call parent constructor
        self.breed = breed

    def fetch(self) -> str:
        return f"{self.name} fetches the ball!"

# Cat inherits from Animal
class Cat(Animal):
    def __init__(self, name: str):
        super().__init__(name, sound="Meow!")

    def purr(self) -> str:
        return f"{self.name} purrs..."

print("--- Basic Inheritance ---")
dog = Dog("Rex", "Labrador")
cat = Cat("Whiskers")

print(f"  {dog.speak()}")               # Rex says Woof!
print(f"  {dog.fetch()}")               # Rex fetches the ball!
print(f"  {cat.speak()}")               # Whiskers says Meow!
print(f"  {cat.purr()}")                # Whiskers purrs...

# isinstance and issubclass checks
print(f"\n  isinstance(dog, Dog)    \u2192 {isinstance(dog, Dog)}")
print(f"  isinstance(dog, Animal) \u2192 {isinstance(dog, Animal)}")
print(f"  issubclass(Dog, Animal) \u2192 {issubclass(Dog, Animal)}")

# Polymorphism — same method, different behavior
print("\n--- Polymorphism ---")
animals = [Dog("Buddy", "Poodle"), Cat("Luna"), Dog("Max", "Beagle")]
for animal in animals:
    print(f"  {animal.info()} \u2192 {animal.speak()}")

# Multi-level inheritance
print("\n--- Multi-level Inheritance ---")
class Puppy(Dog):
    def __init__(self, name: str, breed: str, toy: str):
        super().__init__(name, breed)
        self.toy = toy

    def play(self) -> str:
        return f"{self.name} plays with {self.toy}"

p = Puppy("Tiny", "Chihuahua", "squeaky bone")
print(f"  {p.speak()}")                 # Tiny says Woof!
print(f"  {p.fetch()}")                 # Tiny fetches the ball!
print(f"  {p.play()}")                  # Tiny plays with squeaky bone
```

## 3.3 TYPE HINTS

Type hints add documentation about expected types.
- They don't enforce types at runtime — Python is still dynamically typed.
- Tools like mypy can check type hints statically.
- Available since Python 3.5, improved in 3.9+ and 3.10+.

```python
from typing import Optional, Union

# Basic function type hints
def add(a: int, b: int) -> int:
    return a + b

def greet(name: str, times: int = 1) -> str:
    return (f"Hello, {name}! " * times).strip()

print("--- Function Type Hints ---")
print(f"  add(3, 5)          = {add(3, 5)}")
print(f"  greet('Alice')     = {greet('Alice')!r}")
print(f"  greet('Bob', 2)    = {greet('Bob', 2)!r}")

# Collection type hints (Python 3.9+)
def average(numbers: list[float]) -> float:
    return sum(numbers) / len(numbers)

def count_words(text: str) -> dict[str, int]:
    words = text.lower().split()
    return {w: words.count(w) for w in set(words)}

print(f"\n  average([10, 20, 30]) = {average([10, 20, 30])}")
print(f"  count_words('the cat sat on the mat') = {count_words('the cat sat on the mat')}")

# Optional — value can be the type OR None
def find_user(user_id: int) -> Optional[str]:
    users = {1: "Alice", 2: "Bob"}
    return users.get(user_id)        # Returns str or None

print(f"\n--- Optional ---")
print(f"  find_user(1) = {find_user(1)!r}")
print(f"  find_user(9) = {find_user(9)!r}")    # None

# Union — value can be one of multiple types
def process(value: Union[int, str]) -> str:
    if isinstance(value, int):
        return f"Number: {value * 2}"
    return f"String: {value.upper()}"

print(f"\n--- Union ---")
print(f"  process(5)       = {process(5)!r}")
print(f"  process('hello') = {process('hello')!r}")

# Type hints are NOT enforced at runtime
print(f"\n--- Type Hints Are Not Enforced ---")
result = add("hello", " world")      # Works! No runtime error
print(f"  add('hello', ' world') = {result!r}  (no error at runtime)")
print(f"  But mypy would flag this as a type error.")
```

## 3.4 DATACLASSES

Dataclasses (Python 3.7+) auto-generate `__init__`, `__repr__`, `__eq__`.
- Perfect for classes that mainly store data.
- Use `@dataclass` decorator on the class.
- Use `field()` for mutable default values (like lists).

```python
@dataclass
class Product:
    name: str
    price: float
    quantity: int = 0          # Default value

    def total_value(self) -> float:
        return self.price * self.quantity

print("--- Basic Dataclass ---")
p1 = Product("Laptop", 999.99, 5)
p2 = Product("Laptop", 999.99, 5)
p3 = Product("Mouse", 29.99)

print(f"  p1 = {p1}")                         # Auto-generated __repr__
print(f"  p3 = {p3}")                         # quantity defaults to 0
print(f"  p1 == p2 \u2192 {p1 == p2}")             # Auto-generated __eq__ \u2192 True
print(f"  p1.total_value() = ${p1.total_value():,.2f}")

# Frozen dataclass — immutable (cannot change fields after creation)
@dataclass(frozen=True)
class Point:
    x: float
    y: float

print("\n--- Frozen (Immutable) Dataclass ---")
pt = Point(3.0, 4.0)
print(f"  pt = {pt}")
try:
    pt.x = 99.0              # This will fail!
except AttributeError as e:
    print(f"  Cannot modify frozen: {e}")

# field(default_factory=...) for mutable defaults
@dataclass
class ShoppingCart:
    owner: str
    items: list[str] = field(default_factory=list)  # MUST use field() for lists!

    def add_item(self, item: str):
        self.items.append(item)

    def summary(self) -> str:
        return f"{self.owner}'s cart: {', '.join(self.items) or 'empty'}"

print("\n--- Dataclass with field(default_factory) ---")
cart1 = ShoppingCart("Alice")
cart2 = ShoppingCart("Bob")

cart1.add_item("Book")
cart1.add_item("Pen")
cart2.add_item("Laptop")

print(f"  {cart1.summary()}")                 # Alice's cart: Book, Pen
print(f"  {cart2.summary()}")                 # Bob's cart: Laptop
print(f"  Carts independent: {cart1.items != cart2.items}")  # True

# Ordered dataclass — supports comparison operators
@dataclass(order=True)
class Student:
    gpa: float
    name: str = field(compare=False)   # Don't use name for comparison

print("\n--- Ordered Dataclass ---")
students = [Student(3.5, "Alice"), Student(3.9, "Bob"), Student(3.2, "Charlie")]
students.sort()
for s in students:
    print(f"  GPA {s.gpa} \u2014 {s.name}")
```

## 3.5 `@staticmethod` AND `@property`

- `@staticmethod` — a method that belongs to the class but doesn't access `self` or `cls`. It's like a regular function namespaced under the class.
- `@property` — lets you access a method like an attribute. You can add validation with `@x.setter`. Great for computed attributes.

```python
class Circle:
    def __init__(self, radius: float):
        self._radius = radius        # Convention: _ = "protected"

    # @property — getter: access radius like an attribute
    @property
    def radius(self) -> float:
        return self._radius

    # @radius.setter — add validation when setting
    @radius.setter
    def radius(self, value: float):
        if value < 0:
            raise ValueError("Radius cannot be negative!")
        self._radius = value

    # Read-only computed property (no setter)
    @property
    def area(self) -> float:
        return 3.14159 * self._radius ** 2

    @property
    def circumference(self) -> float:
        return 2 * 3.14159 * self._radius

    # @staticmethod — utility function, no self/cls needed
    @staticmethod
    def is_valid_radius(value) -> bool:
        return isinstance(value, (int, float)) and value >= 0

    def __str__(self):
        return f"Circle(r={self._radius})"

print("--- @property (Getter) ---")
c = Circle(5)
print(f"  c.radius         = {c.radius}")          # Uses getter
print(f"  c.area           = {c.area:.2f}")         # Computed property
print(f"  c.circumference  = {c.circumference:.2f}")

print("\n--- @property (Setter with validation) ---")
c.radius = 10                                       # Uses setter
print(f"  After c.radius = 10:")
print(f"  c.area = {c.area:.2f}")

try:
    c.radius = -3                                   # Triggers validation!
except ValueError as e:
    print(f"  c.radius = -3 \u2192 ValueError: {e}")

print("\n--- @staticmethod ---")
# Called on the class — no instance needed
print(f"  Circle.is_valid_radius(5)     \u2192 {Circle.is_valid_radius(5)}")
print(f"  Circle.is_valid_radius(-1)    \u2192 {Circle.is_valid_radius(-1)}")
print(f"  Circle.is_valid_radius('abc') \u2192 {Circle.is_valid_radius('abc')}")

# Can also be called on an instance (but doesn't use self)
print(f"  c.is_valid_radius(10)         \u2192 {c.is_valid_radius(10)}")

# Another example: @classmethod vs @staticmethod
print("\n--- @classmethod (bonus) ---")

class Temperature:
    def __init__(self, celsius: float):
        self.celsius = celsius

    @property
    def fahrenheit(self) -> float:
        return self.celsius * 9 / 5 + 32

    # @classmethod — alternative constructor
    @classmethod
    def from_fahrenheit(cls, f: float):
        return cls((f - 32) * 5 / 9)

    @staticmethod
    def is_boiling(celsius: float) -> bool:
        return celsius >= 100

    def __str__(self):
        return f"{self.celsius:.1f}\u00b0C / {self.fahrenheit:.1f}\u00b0F"

t1 = Temperature(100)
t2 = Temperature.from_fahrenheit(212)     # classmethod as factory
print(f"  t1 = {t1}")
print(f"  t2 = {t2}")
print(f"  Is 100\u00b0C boiling? {Temperature.is_boiling(100)}")
```

## 3.6 ABSTRACT CLASSES

Abstract classes define a contract — subclasses MUST implement abstract methods.
- Use `ABC` (Abstract Base Class) from the `abc` module.
- You cannot instantiate an abstract class directly.
- Great for defining interfaces and ensuring consistent APIs.

```python
class Shape(ABC):
    # Abstract methods — MUST be implemented by subclasses
    @abstractmethod
    def area(self) -> float:
        pass

    @abstractmethod
    def perimeter(self) -> float:
        pass

    # Concrete method — shared by all subclasses (not abstract)
    def describe(self) -> str:
        return (f"{self.__class__.__name__}: "
                f"area={self.area():.2f}, perimeter={self.perimeter():.2f}")

# Trying to instantiate abstract class
print("--- Cannot Instantiate Abstract Class ---")
try:
    s = Shape()
except TypeError as e:
    print(f"  Shape() \u2192 TypeError: {e}")

# Concrete subclass: Rectangle
class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height

    def perimeter(self) -> float:
        return 2 * (self.width + self.height)

# Concrete subclass: Circle
class CircleShape(Shape):
    def __init__(self, radius: float):
        self.radius = radius

    def area(self) -> float:
        return 3.14159 * self.radius ** 2

    def perimeter(self) -> float:
        return 2 * 3.14159 * self.radius

# Concrete subclass: Triangle
class Triangle(Shape):
    def __init__(self, a: float, b: float, c: float):
        self.a, self.b, self.c = a, b, c

    def area(self) -> float:
        s = (self.a + self.b + self.c) / 2          # Heron's formula
        return (s * (s-self.a) * (s-self.b) * (s-self.c)) ** 0.5

    def perimeter(self) -> float:
        return self.a + self.b + self.c

print("\n--- Polymorphism with Abstract Base Class ---")
shapes: list[Shape] = [
    Rectangle(4, 6),
    CircleShape(5),
    Triangle(3, 4, 5)
]

for shape in shapes:
    print(f"  {shape.describe()}")
# Rectangle: area=24.00, perimeter=20.00
# CircleShape: area=78.54, perimeter=31.42
# Triangle: area=6.00, perimeter=12.00

# What happens if you don't implement all abstract methods?
print("\n--- Missing Implementation ---")
try:
    class IncompleteShape(Shape):
        def area(self) -> float:
            return 0
        # Missing perimeter()!

    obj = IncompleteShape()
except TypeError as e:
    print(f"  IncompleteShape() \u2192 TypeError: {e}")

# Abstract property
print("\n--- Abstract Property ---")

class Vehicle(ABC):
    @property
    @abstractmethod
    def fuel_type(self) -> str:
        pass

    @abstractmethod
    def start(self) -> str:
        pass

class ElectricCar(Vehicle):
    @property
    def fuel_type(self) -> str:
        return "Electric"

    def start(self) -> str:
        return "Silent start..."

class GasCar(Vehicle):
    @property
    def fuel_type(self) -> str:
        return "Gasoline"

    def start(self) -> str:
        return "Vroom!"

cars = [ElectricCar(), GasCar()]
for car in cars:
    print(f"  {car.__class__.__name__}: fuel={car.fuel_type}, start={car.start()}")
```
